```python
import pandas as pd
import json
import re
```

```python
file_path = "/content/fiqa_formatted.json"

with open(file_path, "r") as f:
    data = json.load(f)

# Convert to DataFrame
df = pd.DataFrame(data)

print("Dataset Shape:", df.shape)
df.head()
```

```python
# Keep only required columns
df = df[["question", "answer"]]
```

```python
df = df.dropna()
print("After removing nulls:", df.shape)
```

```python
df = df.dropna()
print("After removing nulls:", df.shape)
```

```python
df = df.drop_duplicates()
print("After removing duplicates:", df.shape)
```

```python
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"http\S+", "", text)
    text = re.sub(r"[^a-zA-Z0-9\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text
```

```python
df["question"] = df["question"].apply(clean_text)
df["answer"] = df["answer"].apply(clean_text)
```

```python
df = df[
    (df["question"].str.len() > 10) &
    (df["answer"].str.len() > 20) &
    (df["answer"].str.len() < 500)
]

print("After filtering:", df.shape)
```

```python
df = df[df["question"] != df["answer"]]
df = df[~df["answer"].str.contains(r"(.)\1{5,}", regex=True)]

print("After quality filtering:", df.shape)
```

```python
df = df.reset_index(drop=True)
df["id"] = df.index
```

```python
output_path = "/content/fiqa_processed.json"

df.to_json(output_path, orient="records", indent=4)

print("Cleaned JSON saved at:", output_path)
```
